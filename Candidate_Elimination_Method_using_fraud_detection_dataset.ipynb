```python
# Candidate Elimination Algorithm
# Dataset: E-Commerce Fraud Detection

# Attributes:
# TransactionAmt, ForeignIP, CardPresent
# Target: Fraud (Yes) / Not Fraud (No)

data = [
    ["High", "Yes", "No",  "Yes"],
    ["High", "Yes", "Yes", "No"],
    ["Low",  "Yes", "No",  "No"],
    ["High", "No",  "No",  "No"],
    ["High", "Yes", "No",  "Yes"],
    ["Low",  "No",  "Yes", "No"],
    ["High", "Yes", "No",  "Yes"],
    ["Low",  "Yes", "No",  "No"]
]

# Attribute domains

domains = [
    ["High", "Low"],
    ["Yes", "No"],
    ["Yes", "No"]
]

# Step 1: Initialize S and G boundaries

S = ["Ø", "Ø", "Ø"]
G = [["?", "?", "?"]]

print("Initial Specific Boundary S:", S)
print("Initial General Boundary G:", G)
print()

# Function to check whether a hypothesis covers an example

def covers(h, example):

    for i in range(len(h)):

        if h[i] != "?" and h[i] != example[i]:
            return False

    return True


# Function to check whether h1 is more general than h2

def more_general(h1, h2):

    for i in range(len(h1)):

        if h1[i] != "?" and h1[i] != h2[i]:
            return False

    return True


# Step 2: Process each training example

for i, example in enumerate(data):

    attributes = example[:-1]
    target = example[-1]

    print("------------------------------------")
    print(f"Processing D{i + 1}: {attributes}")
    print("Target:", target)

    # Positive example

    if target == "Yes":

        print("Positive Example")

        # Generalize S

        if S == ["Ø", "Ø", "Ø"]:

            S = attributes.copy()

        else:

            for j in range(len(S)):

                if S[j] != attributes[j]:
                    S[j] = "?"

        # Remove hypotheses from G that do not cover positive example

        G = [g for g in G if covers(g, attributes)]

        print("Updated S:", S)
        print("Updated G:", G)

    # Negative example

    else:

        print("Negative Example")

        new_G = []

        for g in G:

            if covers(g, attributes):

                # Specialize G to exclude negative example

                for j in range(len(g)):

                    if g[j] == "?":

                        for value in domains[j]:

                            if value != attributes[j]:

                                new_h = g.copy()
                                new_h[j] = value

                                # Keep only hypotheses
                                # that are more general than S

                                if more_general(new_h, S):
                                    new_G.append(new_h)

            else:

                new_G.append(g)

        # Remove duplicate hypotheses

        G = [list(x) for x in set(tuple(h) for h in new_G)]

        # Remove hypotheses that are more specific
        # than another hypothesis in G

        G = [
            g for g in G
            if not any(
                h != g and more_general(h, g)
                for h in G
            )
        ]

        print("Updated S:", S)
        print("Updated G:", G)

    print()


# Step 3: Display final boundaries

print("====================================")
print("FINAL CANDIDATE ELIMINATION RESULT")
print("====================================")

print("Final Specific Boundary S:", S)
print("Final General Boundary G:", G)

print("\nFinal Hypothesis Space:")

for h in G:
    print(h)
```
